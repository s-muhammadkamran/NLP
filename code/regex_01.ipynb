```python
import re
```

```python
chats = [
    "What is your mobile number? Mine is 1234567890.",
    "My Phone number is 222-196-7722.",
    "I've another phone number (333)-196-7722.",
]
```

```python
pattern = r'\d{10}|\d{3}-\d{3}-\d{4}|\(\d{3}\)-\d{3}-\d{4}'
for chat in chats:
    match = re.search(pattern, chat)
    if match:
        print(match.group())
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
txt = "Hello world. This is a test. NLTK is great for NLP tasks."
```

```python
sents = sent_tokenize(txt)
for i, sent in enumerate(sents, start=1):
    print(f"\nSentence {i}: {sent}")
    words = word_tokenize(sent)
    for word in words:
        print(lemmatizer.lemmatize(word))
```

# Spacy

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp("Apple is looking at buying UK. startup for $1 billion. The company is based in California.")
for sent in doc.sents:
    print(sent.text)
```

```python
for sent in doc.sents:
    for token in sent:
        print(token.text, token.pos_, token.dep_)
```
