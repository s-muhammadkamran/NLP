## Performing Airthmatics on Word Vectors

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
king  = nlp.vocab["king"].vector
man   = nlp.vocab["man"].vector
queen = nlp.vocab['queen'].vector
woman = nlp.vocab['woman'].vector

result = king - man + woman
```

### Cosine Similarity

```python
cosine_similarity([result], [queen])
```

```python
doc = nlp("Supply and install reinforced concrete slab 20MPa")
```

```python
x = doc._.trf_data.last_hidden_layer_state
print("type:", type(x))

# try to understand structure
if hasattr(x, "shape"):
    print("shape:", x.shape)
else:
    try:
        print("len:", len(x))
        print("type(x[0]):", type(x[0]))
        if hasattr(x[0], "shape"):
            print("x[0].shape:", x[0].shape)
    except Exception as e:
        print("cannot inspect sequence:", e)

print("repr:", repr(x)[:500])
```

```python
def ragged_mean(ragged) -> np.ndarray:
    """
    ragged: thinc.types.Ragged
    - ragged.data: (sum_lengths, width)
    - ragged.lengths: (batch_size,)
    For a single doc, batch_size == 1 so mean over ragged.data is OK.
    """
    X = ragged.data
    # X could be cupy/torch/np; np.asarray is fine here because it's rectangular
    X = np.asarray(X)
    if X.size == 0:
        raise ValueError("Ragged.data is empty. No transformer token states produced.")
    return X.mean(axis=0)

def sentence_vector(text: str) -> np.ndarray:
    doc = nlp(text)
    td = doc._.trf_data
    # This is a thinc Ragged in your environment
    return ragged_mean(td.last_hidden_layer_state)

def cosine(a: np.ndarray, b: np.ndarray) -> float:
    denom = (np.linalg.norm(a) * np.linalg.norm(b))
    return float(a @ b / denom) if denom else 0.0
```

```python
s1 = "Supply and install reinforced concrete slab 20MPa"
s2 = "Installation of 20 MPa RCC slab"

v1 = sentence_vector(s1)
v2 = sentence_vector(s2)

print("v shape:", v1.shape)
print("similarity:", cosine(v1, v2))  # usually (1, seq_len, hidden), e.g. (1, 20, 768)
```

## Batch embedding (many texts) + top-k similarity

```python
def ragged_means(ragged) -> np.ndarray:
    """
    Compute mean pooling per sequence in a Ragged batch.
    Returns: (batch_size, width)
    """
    data = np.asarray(ragged.data)          # (sum_len, width)
    lengths = np.asarray(ragged.lengths)    # (batch_size,)
    if data.size == 0:
        raise ValueError("Ragged.data is empty.")

    out = np.zeros((len(lengths), data.shape[1]), dtype=data.dtype)

    start = 0
    for i, L in enumerate(lengths):
        end = start + int(L)
        if end > start:
            out[i] = data[start:end].mean(axis=0)
        else:
            out[i] = 0
        start = end
    return out

def embed_texts(texts, batch_size=16) -> np.ndarray:
    vecs = []
    for doc in nlp.pipe(texts, batch_size=batch_size):
        ragged = doc._.trf_data.last_hidden_layer_state
        # This ragged is usually batch_size=1 per doc in nlp.pipe iteration,
        # but we'll handle both safely:
        m = ragged_means(ragged)  # (1, width) typically
        vecs.append(m[0])
    return np.vstack(vecs)        # (n_texts, width)

def top_k(query: str, candidates: list[str], k: int = 5):
    q = embed_texts([query])[0]
    C = embed_texts(candidates)

    q = q / (np.linalg.norm(q) + 1e-12)
    C = C / (np.linalg.norm(C, axis=1, keepdims=True) + 1e-12)

    sims = C @ q
    idx = np.argsort(-sims)[:k]
    return [(candidates[i], float(sims[i])) for i in idx]
```

```python
boq = "Supply and install GI pipe 50mm diameter"
cats = [
    "Plumbing piping systems",
    "Concrete structural works",
    "Electrical conduit installation",
    "Steel framing systems",
    "HVAC air distribution ductwork"
]

for c, s in top_k(boq, cats, k=3):
    print(f"{s:.4f}  {c}")
```
