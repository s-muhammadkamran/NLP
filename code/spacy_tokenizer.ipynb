# Spacy Tokenizer

```python
import spacy
```

# Blank pipline

```python
# Load the English model
nlp = spacy.blank("en")
```

```python
nlp.pipeline
```

```python
doc = nlp('''Spacy is a great library for natural language processing. "Let's go to N.Y.!"''')

for token in doc:
    print(f"\nToken:{token.text}, is_alpha:{token.is_alpha}, is_stop:{token.is_stop}, pos:{token.pos_}")
    print(f"dep:{token.dep_}, lemma:{token.lemma_}, pos:{token.pos_}, tag:{token.tag_}, dep:{token.dep_}")
```

# Prebuilt English pipline

```python
nlp = spacy.load("en_core_web_sm")
```

```python
nlp.pipeline
```

```python
doc = nlp('''Spacy is a great library for natural language processing. "Let's go to N.Y.!"''')
for token in doc:
    print(f"\nToken:{token.text}, is_alpha:{token.is_alpha}, is_stop:{token.is_stop}, pos:{token.pos_}")
    print(f"dep:{token.dep_}, lemma:{token.lemma_}, pos:{token.pos_}, tag:{token.tag_}, dep:{token.dep_}")
```

```python
# Generate a unique long text string having 1000 words and add a few urls in it as well
long_text = " ".join([f"This is a sample sentence {i}." for i in range(1000)]) + " Visit https://www.example.com for more information. Check out http://www.test.com as well."
```

```python
doc1 = nlp(long_text)

urls = []
for token in doc1:
    if token.like_url:
        urls.append(token.text)

print(f"\nURLs found in the text: {', '.join(urls)}")
```

```python
from spacy import displacy
```

```python
displacy.render(doc, style="ent")
```

# Custom Pipeline

```python
source_nlp = spacy.load("en_core_web_sm")
target_nlp = spacy.blank("en")
```

```python
target_nlp.add_pipe("ner", source=source_nlp)
target_nlp.pipe_names
```

```python
doc = nlp("Tesla is a company founded by Elon Musk and it is going to acquire Twitter for $44 billion.")
```

```python
print(f"{'Token':^12}", " | ", f"{'Label':^10}", " | ", f"{'Explanation':^20}")
print("=" * 14 + "|" + "=" * 14 + "|" + "=" * 41)
for ent in doc.ents:    
    print(f"{ent.text:^12}", " | ", f"{ent.label_:^10}", " | ", spacy.explain(ent.label_))
```
